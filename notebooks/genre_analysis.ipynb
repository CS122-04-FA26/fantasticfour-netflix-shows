```python
from collections import Counter
from itertools import combinations

from load_data import load_netflix_data


def get_genres(listed_in):
    """Turn 'Dramas, Romantic Movies' into ['Dramas', 'Romantic Movies']."""
    return [genre.strip() for genre in listed_in.split(",")]


def count_genre_pairs(df):
    """Count how often each pair of genres appears together on a title."""
    pair_counts = Counter()
    for listed_in in df["listed_in"].dropna():
        genres = sorted(get_genres(listed_in))
        for pair in combinations(genres, 2):
            pair_counts[pair] += 1
    return pair_counts


if __name__ == "__main__":
    df = load_netflix_data()
    for (g1, g2), count in count_genre_pairs(df).most_common(10):
        print(f"{g1} + {g2}: {count}")
```
